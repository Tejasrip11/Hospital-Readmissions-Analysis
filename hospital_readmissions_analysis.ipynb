```python
import pandas as pd

file_path = "data/FY_2026_Hospital_Readmissions_Reduction_Program_Hospital.csv"
df = pd.read_csv(file_path, dtype={"Facility ID": "string"})

print("Rows and columns:", df.shape)
df.head()
```

```python
df.info()
print("\nMissing values by column:")
print(df.isna().sum())
```

```python
print("Footnote values:")
print(df["Footnote"].value_counts(dropna=False))

print("\nReadmissions values that are not numbers:")
values = df["Number of Readmissions"].dropna().astype(str)
print(values[pd.to_numeric(values, errors="coerce").isna()].value_counts())
```
